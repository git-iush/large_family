```python
import pandas as pd
import numpy as np
```

```python
url = 'https://github.com/git-iush/large_family/raw/refs/heads/main/nhis_00002.csv.gz'
df = pd.read_csv(url)
```

```python
df.head()
```

```python
missing_codes = {
    # 3-digit variables
    "AGE": [997, 998, 999], "RACENEW": [997, 998, 999],
    "EDUC": [997, 998, 999], "DVINT": [997, 998, 999],
    # 2-digit variables
    "USBORN": [97, 98, 99], "ARMFEV": [97, 98, 99], "OWNERSHIP": [97, 98, 99],
    "HOURSWRK": [97, 98, 99], "DVINTWELL": [97, 98, 99],
    "GOTWELF": [70, 80, 90], "GOTINT": [70, 80, 90], "GOTSTAMPFAM": [70, 80, 90],
}
# IDs, weights, and variables with no missing codes: leave alone
skip = set(missing_codes) | {"YEAR","SERIAL","STRATA","PSU","NHISHID","REGION","URBRRL",
        "PERNUM","NHISPID","HHX","SAMPWEIGHT","ASTATFLG","CSTATFLG","CPI2009",
        "FAMSIZECAT","POVERTY","HISPYN"}
# Everything else is a 1-digit yes/no style code: 7/8/9 = missing
for col in df.columns:
    if col not in skip:
        missing_codes[col] = [7, 8, 9]

df = df.replace(missing_codes, np.nan)
```

```python
nan_counts = df.isna().sum()
print(nan_counts[nan_counts > 0].sort_values(ascending=False))
print("Rows with any NaN:", df.isna().any(axis=1).sum())
```

```python
nan_pct = df.isna().mean() * 100
```

```python
print(nan_pct[nan_pct > 5].sort_values(ascending=False).round(2))
```
