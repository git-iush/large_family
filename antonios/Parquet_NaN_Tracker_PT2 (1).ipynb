```python
df = pd.read_parquet("/Users/antonioskoskinas/Downloads/data.parquet")
```

```python
df.head()
```

```python
df = df.drop(columns=['POVERTY','MOVEDIN'])
```

```python
# Project sample: women, all ages
df = df[df["SEX"] == 2].copy()
print("Women:", len(df))
```

```python
df.head()
```

```python
#Missing Codes
missing_codes = {
    # Income variables
    "INCTOT":   [9999999],
    "FTOTINC":  [9999998, 9999999],
    "INCWELFR": [99999],
    # 0 = N/A
    "EMPSTAT":  [0],
    "UHRSWORK": [0],
}

for col, codes in missing_codes.items():
    df[col] = df[col].replace(codes, np.nan)
```

```python
# Project sample is women 18-45 and older
df = df[(df["SEX"] == 2) & (df["AGE"].between(18, 45))].copy()
print("Women 18-45:", len(df))
```

```python
nan_counts = df.isna().sum()
print(nan_counts[nan_counts > 0].sort_values(ascending=False))
print("Rows with any NaN:", df.isna().any(axis=1).sum())
```

```python
nan_pct = df.isna().mean() * 100
print(nan_pct.sort_values(ascending=False).round(2))
```
